## Zmienne

Zmienna w Pythonie to nazwany *„pojemnik”* służący do przechowywania danych w pamięci komputera. Wystarczy podać jej nazwę i przypisać wartość za pomocą znaku równości =. Python jest językiem dynamicznie typowanym, więc nie musisz określać typu zmiennej podczas jej tworzenia.

**Typy zmiennych:**

- str
- bool
- int
- float

**Przykład**
```
wiek = 16
imie = 'Tomek'
wzrost = 1.70
logiczna = True
```





````python
wiek = 16
imie = 'Tomek'
wzrost = 1.70
logiczna = True

print(f'Mam na imię {imie}, mam {wiek} lat, mój wzrost to {wzrost} m, Czy lubię chodzić do kina ? {logiczna}')
````
